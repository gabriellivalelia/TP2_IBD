# Análise da Expansão da Rede de Transmissão de Energia Elétrica Brasileira a partir de Dados Abertos da ANEEL

>A expansão da rede básica de transmissão de energia elétrica tem seu planejamento centralizado. O Ministério de Minas e Energia — MME, a partir dos estudos técnicos elaborados pela Empresa de Pesquisa Energética — EPE e pelo Operador Nacional do Sistema Elétrico — ONS, consolida os programas de outorgas, observando os preceitos técnicos e de modicidade tarifária.
>
>Definidos os Programas de Expansão pelo MME, a Agência Nacional de Energia Elétrica (ANEEL) operacionaliza a concessão de novas outorgas, observando a Lei 8.987/1995.
>
>As outorgas de transmissão têm seus dados cadastrais atualizados pela Superintendência de Concessões, Permissões e Autorizações de Transmissão e Distribuição (SCT); sua fase de implantação monitorada pela Superintendência de Fiscalização dos Serviços de Eletricidade (SFE); e a gestão de receitas pós processo de reajuste gerida pela Superintendência de Gestão Tarifária (SGT), por meio do Sistema de Gestão da Transmissão — SIGET.
>
>O monitoramento da expansão tem como objetivo o acompanhamento contínuo da execução das obras de empreendimentos de transmissão e a produção de informações relevantes sobre a expansão do sistema de transmissão.
>
>O Relatório de Monitoramento de Empreendimentos de Transmissão é divulgado mensalmente e detalha as obras e marcos de execução dos empreendimentos em andamento ou concluídos recentemente. Entre as informações apresentadas estão a data de entrada em operação estabelecida no ato legal, a data de conclusão informada pela transmissora e a situação do empreendimento, segundo informações atualizadas no SIGET.
>
>As datas de previsão apresentadas no relatório de monitoramento são informadas mensalmente pelas transmissoras por meio do Siget. Não há interferência direta da SFE na previsão publicada mensalmente, ou seja, as informações apresentadas no relatório são de responsabilidade das concessionárias de transmissão. Por outro lado, o monitoramento atua no sentido de garantir a qualidade das informações prestadas pelas transmissoras.
>
>Adicionalmente, os empreendimentos de transmissão de maior relevância para o Sistema Interligado Nacional (SIN) passam por um acompanhamento diferenciado, que objetiva manter um conjunto detalhado de informações sobre o andamento das obras de implantação desses empreendimentos. O resultado desse acompanhamento é consolidado no Relatório Trimestral de Acompanhamento Diferenciado de Empreendimentos de Transmissão.

# Descrição e Manipulação dos Dados

## Origem e caracterização da base de dados

O conjunto de dados utilizado neste trabalho foi obtido do **Sistema de Gestão da Transmissão (SIGET)**, disponibilizado pela **Agência Nacional de Energia Elétrica (ANEEL)** no Portal Brasileiro de Dados Abertos.

O SIGET reúne informações referentes à gestão, à implantação e à expansão das instalações de transmissão de energia elétrica no Brasil. Os dados abrangem aspectos administrativos e técnicos do setor, incluindo agentes responsáveis, contratos de concessão, empreendimentos, obras, módulos de transmissão, linhas de transmissão, subestações e equipamentos elétricos.

A utilização dessa base permite investigar a organização e a expansão da infraestrutura de transmissão de energia elétrica brasileira, relacionando informações sobre os responsáveis pelos empreendimentos às características técnicas das instalações.

**Fontes dos dados:**

- **Órgão responsável:** Agência Nacional de Energia Elétrica (ANEEL).
- **Sistema de origem:** Sistema de Gestão da Transmissão (SIGET).
- **Domínio:** Transmissão de energia elétrica e expansão do Sistema Interligado Nacional (SIN).
- **Conjunto de dados SIGET:** https://dados.gov.br/dados/conjuntos-dados/sistema-de-gestao-da-transmissao-siget

Os arquivos utilizados foram disponibilizados em formato CSV e apresentam data de geração de 24 de outubro de 2026. A documentação da base também inclui dicionários de dados oficiais, utilizados para identificar e interpretar os atributos disponíveis.

## Processamento e normalização dos dados

O processamento dos dados foi realizado em Python, utilizando a biblioteca **Pandas** para leitura, análise e transformação dos arquivos CSV. A implementação do banco de dados relacional foi realizada com o **SQLite**, por meio da biblioteca `sqlite3`.

Seguindo a abordagem *bottom-up* proposta no enunciado, a modelagem partiu da estrutura dos arquivos originais e da identificação de entidades, atributos, dependências funcionais e relacionamentos existentes nos dados.

Inicialmente, os cinco arquivos foram carregados e analisados quanto à quantidade de registros, atributos disponíveis, identificadores, valores ausentes e possíveis inconsistências.

Em seguida, foram identificados os atributos que descrevem diferentes entidades, permitindo a decomposição das estruturas originais em relações específicas. Esse procedimento buscou reduzir a redundância de informações e evitar anomalias de inserção, atualização e exclusão.

### Estrutura das relações normalizadas

A seguir, são apresentadas as relações que compõem o banco de dados, com suas respectivas finalidades e atributos. As descrições consideram a organização estabelecida durante o processo de normalização.

As indicações **PK** (*Primary Key*) e **FK** (*Foreign Key*) identificam, respectivamente, as chaves primárias e estrangeiras.

#### a) Agente

A relação `Agente` armazena informações cadastrais das empresas responsáveis pelos contratos de transmissão de energia elétrica. Sua identificação é realizada pelo campo `IdAgente`.

| Atributo | Descrição |
|---|---|
| `IdAgente` (PK) | Identificador único do agente |
| `NumCNPJ` | Número do CNPJ do agente |
| `DscRazaoSocial` | Razão social da empresa |
| `NumCEPAgente` | Código de Endereçamento Postal |
| `SigUF` | Sigla da unidade federativa |
| `NomMunicipio` | Nome do município |
| `NomBairroAgente` | Nome do bairro |
| `DscEnderecoAgente` | Endereço cadastral do agente |

A separação dessas informações evita a repetição dos dados cadastrais de um mesmo agente em diferentes contratos.

#### b) Contrato

A relação `Contrato` armazena informações dos instrumentos de concessão ou autorização associados aos agentes de transmissão.

| Atributo | Descrição |
|---|---|
| `IdeCcd` (PK) | Identificador único do contrato |
| `IdAgente` (FK) | Identificador do agente responsável |
| `IdcTipoCcd` | Indicador do tipo de contrato |
| `NumCnaCcd` | Número de identificação do contrato |
| `DatAsnCcd` | Data de assinatura do contrato |
| `DatFimCcd` | Data final do contrato de concessão |
| `IdeCcoTarReceita` | Identificador da tarifa de receita associada ao contrato |

O campo `IdAgente` estabelece a associação entre cada contrato e o agente responsável. Durante o processamento, os contratos sem identificação válida do agente foram excluídos.

#### c) Empreendimento

A relação `Empreendimento` representa projetos de implantação ou expansão das instalações de transmissão de energia elétrica.

| Atributo | Descrição |
|---|---|
| `IdeEpd` (PK) | Identificador único do empreendimento |
| `IdeCcd` (FK) | Identificador do contrato associado |
| `IdeOnsEpd` | Identificador do empreendimento no ONS |
| `NomEpd` | Nome do empreendimento |
| `DscEpd` | Descrição do empreendimento |
| `DscSituacaoEpd` | Situação do empreendimento |
| `DatCaoCgmAtoLgl` | Data de previsão de conclusão do empreendimento estabelecida no ato legal |
| `DatOprComEpd` | Data de operação comercial estabelecida no ato legal |
| `DatEfeOprComEpd` | Data efetiva de início da operação comercial |

A relação permite acompanhar a situação dos empreendimentos e comparar datas previstas e efetivas, contribuindo para análises sobre a implantação da infraestrutura de transmissão.

#### d) Obra

A relação `Obra` representa as intervenções relacionadas à execução dos empreendimentos de transmissão. Cada obra está vinculada a um empreendimento e a um módulo.

| Atributo | Descrição |
|---|---|
| `IdeObr` (PK) | Identificador único da obra |
| `IdeEpd` (FK) | Identificador do empreendimento |
| `IdeMdl` (FK) | Identificador do módulo associado |
| `DscObr` | Descrição da obra |
| `DscSitObr` | Situação da obra |
| `DscTipObr` | Tipo de obra |
| `NumVdaUtlMdl` | Vida útil do módulo considerada para a obra |
| `VlrHisRct` | Valor atualizado da receita do último ciclo tarifário |
| `DatOprComObr` | Data de operação comercial associada à obra |

O atributo `NumVdaUtlMdl` foi mantido nessa relação porque seus valores podem variar entre diferentes obras associadas a um mesmo módulo. De maneira semelhante, `VlrHisRct` foi associado à obra conforme a dependência funcional identificada nos dados.

#### e) Modulo

A relação `Modulo` representa as unidades funcionais que compõem as instalações de transmissão de energia elétrica. Ela reúne características comuns aos diferentes tipos de módulos.

A relação concentra os atributos gerais dos módulos, enquanto suas características específicas são distribuídas nas relações `ModuloLinha`, `ModuloEquipamento` e `ModuloManobra`.

### Relacionamentos entre as entidades

Os relacionamentos foram definidos a partir dos identificadores compartilhados nos arquivos e das dependências observadas durante a análise dos dados.

A Tabela 3 apresenta os principais relacionamentos do modelo.

**Tabela 3 — Relacionamentos do banco normalizado.**

| Entidades relacionadas | Cardinalidade | Descrição |
|---|---|---|
| Agente — Contrato | 1:N | Um agente pode possuir vários contratos, enquanto cada contrato possui um agente responsável |
| Contrato — Empreendimento | 1:N | Um contrato pode abranger vários empreendimentos |
| Empreendimento — Obra | 1:N | Um empreendimento pode ser composto por várias obras |
| Modulo — Obra | 1:N | Um módulo pode estar associado a diferentes obras |
| Modulo — ModuloLinha | 1:0..1 | Um módulo pode possuir informações específicas de linha |
| Modulo — ModuloEquipamento | 1:0..1 | Um módulo pode possuir informações específicas de equipamento |
| Modulo — ModuloManobra | 1:0..1 | Um módulo pode possuir informações específicas de manobra |
| LinhaTransmissao — ModuloLinha | 1:N | Uma linha pode estar associada a vários módulos de linha |
| Subestacao — ModuloEquipamento | 1:N | Uma subestação pode conter vários módulos de equipamento |
| Subestacao — ModuloManobra | 1:N | Uma subestação pode conter vários módulos de manobra |
| LinhaTransmissao — Subestacao | M:N | Uma linha possui subestações terminais e uma subestação pode se conectar a várias linhas |

O relacionamento entre `LinhaTransmissao` e `Subestacao` merece destaque, pois representa a conectividade física entre as instalações. Sua implementação foi realizada por meio da tabela associativa `LinhaSubestacao`.

As relações `ModuloLinha`, `ModuloEquipamento` e `ModuloManobra`, por sua vez, representam uma especialização da entidade `Modulo`. Essas relações compartilham sua chave primária com a entidade geral e acrescentam apenas os atributos específicos de cada categoria.

A especialização é considerada parcial, uma vez que nem todos os módulos cadastrados pertencem a uma das três categorias selecionadas. A exclusividade entre as categorias não é imposta diretamente pelas restrições implementadas no SQLite.

## Limpeza e validação dos dados

Antes da inserção dos registros no banco normalizado, foram realizadas verificações de qualidade e consistência dos dados, incluindo:

1. **Identificação de registros duplicados:** análise da repetição de identificadores e distinção entre redundâncias e ocorrências válidas de uma mesma entidade em diferentes registros.
2. **Verificação de dependências funcionais:** análise da consistência dos atributos associados a cada identificador, de modo a evitar a atribuição de valores contraditórios a uma mesma entidade.
3. **Tratamento de valores ausentes:** identificação de atributos obrigatórios sem preenchimento e definição de estratégias de tratamento.
4. **Padronização dos tipos de dados:** conversão de campos numéricos e datas para formatos compatíveis com o SQLite.
5. **Validação de chaves estrangeiras:** conferência da existência dos registros referenciados pelas relações.
6. **Verificação de cardinalidades:** identificação dos relacionamentos entre as entidades, incluindo as associações entre linhas de transmissão e subestações.

Durante o processamento, foram identificadas inconsistências relacionadas à ausência de identificadores, à integridade referencial e à divergência de atributos cadastrais. Para garantir a consistência do banco normalizado, adotou-se como critério a exclusão dos registros inválidos e, quando necessário, dos registros dependentes afetados por essas exclusões.

Inicialmente, foi removido um contrato sem identificação válida do agente responsável. Também foram excluídos 293 registros do arquivo central associados a 14 identificadores distintos de contratos proprietários que não possuíam correspondência no cadastro de contratos utilizado.

Como consequência dessas exclusões, foram identificados registros de módulos especializados sem associação com obras válidas. Dessa forma, foram removidos registros dos arquivos de módulos de linha, equipamentos e manobra que perderam suas referências.

Adicionalmente, foram identificadas seis subestações com valores contraditórios em seus atributos descritivos. Optou-se pela exclusão integral dessas subestações e dos registros que dependiam delas, em vez de selecionar arbitrariamente uma das descrições disponíveis.

# Criação das Tabelas

```python
import pandas as pd
import sqlite3, re
from pathlib import Path
from collections import defaultdict

# Arquivos de entrada: cada chave identifica o conjunto de dados usado no processamento.
ARQUIVOS = {
    "agente": "siget-contrato-agente.csv",
    "central": "siget-contrato-empreendimento-obra-modulo.csv",
    "linha": "siget-contrato-modulolinhatransmissao-subestacaoorigem-subestacaodestino.csv",
    "equip": "siget-contrato-moduloequipamento-subestacao.csv",
    "manobra": "siget-contrato-modulomanobra-subestacao-tipoarranjo.csv",
}
# Diretórios de entrada e saída do processamento.
BASE = Path("data")
SAIDA = Path("resultado_siget")
SAIDA.mkdir(exist_ok=True)
DB = SAIDA / "siget_normalizado.db"
# `logs` resume as etapas; `rejeicoes` armazena os registros removidos.
logs = []
rejeicoes = []


# Registra a quantidade de linhas afetadas em cada etapa para a auditoria.
def log(etapa, tabela, motivo, quantidade):
    logs.append(
        {
            "etapa": etapa,
            "tabela": tabela,
            "motivo": motivo,
            "quantidade": int(quantidade),
        }
    )


# Lê um CSV e padroniza os nomes e os valores textuais antes da limpeza.
def carregar(arq):
    d = pd.read_csv(
        BASE / arq, sep=";", encoding="latin1", dtype="string", low_memory=False
    )
    # Remove espaços dos nomes das colunas e transforma valores vazios em NA.
    d.columns = d.columns.str.strip()
    for c in d:
        d[c] = (
            d[c]
            .str.strip()
            .replace({"": pd.NA, "nan": pd.NA, "null": pd.NA, "NULL": pd.NA})
        )
    return d


# Carrega todos os arquivos em DataFrames independentes, sem alterar os CSVs.
D = {k: carregar(f) for k, f in ARQUIVOS.items()}
for k, v in D.items():
    print(f"{k:8s} {len(v):>7,} registros, {len(v.columns)} colunas")


# Guarda os registros rejeitados e os inclui no relatório de auditoria.
def registrar_rejeicao(tabela, motivo, df):
    if len(df):
        rej = df.copy()
        rej.insert(0, "motivo", motivo)
        rejeicoes.append((tabela, motivo, rej))
        log("rejeicao", tabela, motivo, len(rej))


def extrair(df, tabela, chave, cols, obrigatorias=()):
    """Cria uma relação com uma linha por chave e valida seus atributos."""
    cols = [x for x in dict.fromkeys([chave] + list(cols)) if x in df.columns]
    base = df[cols].copy()
    # Uma chave ausente impede a identificação da entidade e causa rejeição.
    invalid = base[chave].isna()
    registrar_rejeicao(tabela, "chave primária ausente", base.loc[invalid])
    base = base.loc[~invalid]
    # A mesma chave não pode possuir valores diferentes para um mesmo atributo.
    conflitos = set()
    for c in cols:
        if c == chave:
            continue
        n = base.groupby(chave, dropna=False)[c].nunique(dropna=True)
        conflitos.update(n[n > 1].index.astype(str).tolist())
    if conflitos:
        registrar_rejeicao(
            tabela,
            "atributos conflitantes para a mesma chave",
            base[base[chave].isin(conflitos)],
        )
        base = base[~base[chave].isin(conflitos)]

    # O agrupamento combina registros complementares, preservando valores não nulos.
    def primeiro(s):
        s = s.dropna()
        return s.iloc[0] if len(s) else None

    tab = base.groupby(chave, as_index=False, dropna=False).first()
    # Remove entidades que ainda não possuem algum atributo obrigatório.
    for c in obrigatorias:
        if c in tab:
            bad = tab[c].isna()
            registrar_rejeicao(tabela, f"{c} obrigatório ausente", tab[bad])
            tab = tab[~bad]
    log("extracao", tabela, "linhas criadas", len(tab))
    return tab.reset_index(drop=True)


# Mantém somente registros cuja chave estrangeira existe na relação pai.
def filtrar_fk(df, tabela, fk, pais, pk):
    good = df[fk].isin(set(pais[pk].dropna())) & df[fk].notna()
    registrar_rejeicao(tabela, f"FK {fk} sem correspondente em {pk}", df[~good])
    return df[good].reset_index(drop=True)


# Cria cópias de trabalho para preservar os dados originais.
A = D["agente"].copy()
C = D["central"].copy()
L = D["linha"].copy()
E = D["equip"].copy()
M = D["manobra"].copy()

# Limpeza estrita: registros inválidos são excluídos, em vez de mantidos com FK nula.
# As remoções ocorrem somente nos DataFrames em memória; os CSVs não são alterados.
sem_agente = A["IdAgente"].isna()
registrar_rejeicao(
    "Contrato", "Contrato excluído por IdAgente ausente", A.loc[sem_agente]
)
A = A.loc[~sem_agente].copy()
agente = extrair(
    A,
    "Agente",
    "IdAgente",
    [
        "NumCNPJ",
        "DscRazaoSocial",
        "NumCEPAgente",
        "SigUF",
        "NomMunicipio",
        "NomBairroAgente",
        "DscEnderecoAgente",
    ],
)
contrato = extrair(
    A,
    "Contrato",
    "IdeCcd",
    ["IdAgente", "IdcTipoCcd", "NumCnaCcd", "DatAsnCcd", "DatFimCcd"],
    ["IdAgente"],
)
contrato = filtrar_fk(contrato, "Contrato", "IdAgente", agente, "IdAgente")
# O campo tarifário vem do arquivo central e é incorporado à relação Contrato.
# Antes da junção, verifica-se se IdeCcd determina um único IdeCcoTarReceita.
# A data DatFimCcd permanece no cadastro de contratos, pois pode variar no arquivo central.
tar = C[["IdeCcd", "IdeCcoTarReceita"]].dropna(subset=["IdeCcd"]).copy()
conflitos_tar = tar.groupby("IdeCcd")["IdeCcoTarReceita"].nunique(dropna=True)
if (conflitos_tar > 1).any():
    raise ValueError("IdeCcoTarReceita não é funcionalmente dependente de IdeCcd!")
tar = tar.dropna(subset=["IdeCcoTarReceita"]).drop_duplicates("IdeCcd")
contrato = contrato.merge(tar, on="IdeCcd", how="left", validate="one_to_one")
# Valida as referências a contratos proprietários no arquivo central.
# Um valor nulo não é rejeitado; somente uma referência preenchida e inexistente é inválida.
ids_contratos = set(contrato["IdeCcd"].dropna())
if "IdeCcdProprietario" not in C.columns:
    raise KeyError(
        "Arquivo central sem IdeCcdProprietario; não é seguro ignorar a validação."
    )
sem_proprietario = C["IdeCcdProprietario"].notna() & ~C["IdeCcdProprietario"].isin(
    ids_contratos
)
registrar_rejeicao(
    "Central",
    "IdeCcdProprietario não encontrado: linha removida",
    C.loc[sem_proprietario],
)
ids_proprietarios_ausentes = sorted(set(C.loc[sem_proprietario, "IdeCcdProprietario"]))
pd.DataFrame({"IdeCcdProprietario_nao_encontrado": ids_proprietarios_ausentes}).to_csv(
    SAIDA / "contratos_proprietarios_ausentes.csv", index=False
)
C = C.loc[~sem_proprietario].copy()
# Remove as linhas sem contrato principal válido, inclusive as afetadas pela
# exclusão anterior de agentes sem identificador.
sem_contrato = C["IdeCcd"].isna() | ~C["IdeCcd"].isin(ids_contratos)
registrar_rejeicao(
    "Central", "IdeCcd sem contrato válido: linha removida", C.loc[sem_contrato]
)
C = C.loc[~sem_contrato].copy()
empreendimento = extrair(
    C,
    "Empreendimento",
    "IdeEpd",
    [
        "IdeCcd",
        "IdeOnsEpd",
        "NomEpd",
        "DscEpd",
        "DscSituacaoEpd",
        "DatCaoCgmAtoLgl",
        "DatOprComEpd",
        "DatEfeOprComEpd",
    ],
    ["IdeCcd"],
)
empreendimento = filtrar_fk(
    empreendimento, "Empreendimento", "IdeCcd", contrato, "IdeCcd"
)
modulo = extrair(
    C,
    "Modulo",
    "IdeMdl",
    [
        "IdeOnsMdl",
        "NomMdl",
        "DscMdl",
        "IdeTipMdl",
        "DscSituacaoModulo",
        "DscTipMdl",
        "SigTipMdl",
        "DscClfMdl",
        "SglClfMdl",
        "DscItmClfMdl",
    ],
)
obra = extrair(
    C,
    "Obra",
    "IdeObr",
    [
        "IdeEpd",
        "IdeMdl",
        "DscObr",
        "DscSitObr",
        "DscTipObr",
        "NumVdaUtlMdl",
        "VlrHisRct",
        "DatOprComObr",
    ],
    ["IdeEpd", "IdeMdl"],
)
obra = filtrar_fk(obra, "Obra", "IdeEpd", empreendimento, "IdeEpd")
obra = filtrar_fk(obra, "Obra", "IdeMdl", modulo, "IdeMdl")
# Remove módulos e empreendimentos que ficaram sem obras após as validações.
# Assim, nenhuma entidade dependente permanece órfã no banco normalizado.
modulo = modulo.loc[modulo["IdeMdl"].isin(set(obra["IdeMdl"]))].reset_index(drop=True)
empreendimento = empreendimento.loc[
    empreendimento["IdeEpd"].isin(set(obra["IdeEpd"]))
].reset_index(drop=True)
# Os arquivos especializados só podem conservar linhas associadas a módulos válidos.
mods = set(modulo["IdeMdl"])
for nome, df in [("Linha", L), ("Equipamento", E), ("Manobra", M)]:
    invalidos = df["IdeMdl"].isna() | ~df["IdeMdl"].isin(mods)
    registrar_rejeicao(
        nome, "Módulo sem obra válida: linha removida", df.loc[invalidos]
    )
    if nome == "Linha":
        L = df.loc[~invalidos].copy()
    if nome == "Equipamento":
        E = df.loc[~invalidos].copy()
    if nome == "Manobra":
        M = df.loc[~invalidos].copy()


# Subestações: rejeita integralmente identificadores com atributos contraditórios.
# A mesma subestação pode aparecer em diferentes arquivos, mas seus dados cadastrais
# precisam ser consistentes. Não é escolhido um valor arbitrário por maioria.
# Reúne as quatro formas de referência a subestações em uma estrutura única.
def reunir_subestacoes(L, E, M):
    partes = []
    for df, ident, ons, nom, longo, uf in [
        (
            L,
            "IdeSbeOri",
            "IdeOnsSbeOrigem",
            "NomSubestacaoOrigem",
            "NomLongoSubestacaoOrigem",
            "SigUFSubestacaoOrigem",
        ),
        (
            L,
            "IdeSbeDsn",
            "IdeOnsSbeDestino",
            "NomSubestacaoDestino",
            "NomLongoSubestacaoDestino",
            "SigUFSubestacaoDestino",
        ),
        (
            E,
            "IdeSbe",
            "IdeOnsSbe",
            "NomSubestacao",
            "NomLongoSubestacao",
            "SigUFSubestacao",
        ),
        (
            M,
            "IdeSbe",
            "IdeOnsSbe",
            "NomSubestacao",
            "NomLongoSubestacao",
            "SigUFSubestacao",
        ),
    ]:
        temp = df[[ident, ons, nom, longo, uf]].copy()
        temp.columns = [
            "IdeSbe",
            "IdeOnsSbe",
            "NomSubestacao",
            "NomLongoSubestacao",
            "SigUFSubestacao",
        ]
        partes.append(temp)
    return pd.concat(partes, ignore_index=True).dropna(subset=["IdeSbe"])


# Identifica atributos que assumem mais de um valor para o mesmo identificador.
sub_raw = reunir_subestacoes(L, E, M)
atributos_sub = ["IdeOnsSbe", "NomSubestacao", "NomLongoSubestacao", "SigUFSubestacao"]
conflitantes = set()
for campo in atributos_sub:
    q = sub_raw.groupby("IdeSbe")[campo].nunique(dropna=True)
    conflitantes.update(q[q > 1].index.tolist())
if conflitantes:
    registrar_rejeicao(
        "Subestacao",
        "Identificadores com atributos contraditórios; excluídos",
        sub_raw.loc[sub_raw.IdeSbe.isin(conflitantes)].drop_duplicates(),
    )
    registrar_rejeicao(
        "Linha",
        "Terminal vinculado a subestação inconsistente; excluído",
        L.loc[L.IdeSbeOri.isin(conflitantes) | L.IdeSbeDsn.isin(conflitantes)],
    )
    L = L.loc[~(L.IdeSbeOri.isin(conflitantes) | L.IdeSbeDsn.isin(conflitantes))].copy()
    registrar_rejeicao(
        "Equipamento",
        "Subestação inconsistente; registro excluído",
        E.loc[E.IdeSbe.isin(conflitantes)],
    )
    E = E.loc[~E.IdeSbe.isin(conflitantes)].copy()
    registrar_rejeicao(
        "Manobra",
        "Subestação inconsistente; registro excluído",
        M.loc[M.IdeSbe.isin(conflitantes)],
    )
    M = M.loc[~M.IdeSbe.isin(conflitantes)].copy()
    log(
        "limpeza",
        "Subestacao",
        "identificadores de subestação excluídos",
        len(conflitantes),
    )
# Recalcula a base após as exclusões e confirma que nenhuma divergência permaneceu.
sub_raw = reunir_subestacoes(L, E, M)
for campo in atributos_sub:
    q = sub_raw.groupby("IdeSbe")[campo].nunique(dropna=True)
    assert not (q > 1).any(), f"Divergências remanescentes: {campo}"


def primeiro_nao_nulo(s):
    s = s.dropna()
    return s.iloc[0] if len(s) else None


subestacao = sub_raw.groupby("IdeSbe", as_index=False).agg(
    {c: primeiro_nao_nulo for c in atributos_sub}
)
log("extracao", "Subestacao", "linhas válidas criadas", len(subestacao))
linha = extrair(
    L,
    "LinhaTransmissao",
    "IdeLinTms",
    ["NomLinTms", "DscSitLinTms", "NumTensaoBaseLinhaTransm"],
)
modulo_linha = extrair(
    L,
    "ModuloLinha",
    "IdeMdl",
    ["IdeLinTms", "NumCcuLinTms", "QtdTorLinTms", "NumEtnLinTms"],
)
modulo_linha = filtrar_fk(modulo_linha, "ModuloLinha", "IdeMdl", modulo, "IdeMdl")
modulo_linha = filtrar_fk(modulo_linha, "ModuloLinha", "IdeLinTms", linha, "IdeLinTms")
modulo_equip = extrair(
    E,
    "ModuloEquipamento",
    "IdeMdl",
    [
        "IdeSbe",
        "DscTipEqp",
        "NumFasMdlEqp",
        "MdaPotAtvMdlEqp",
        "MdaTnsScdMdlEqp",
        "MdaTnsTiaMdlEqp",
        "DscFinMdlEqp",
        "MdaCpeRaiNegMdlEqp",
        "MdaCpeRaivPosMdlEqp",
        "DscIdcExsMdlCnr",
    ],
)
modulo_equip = filtrar_fk(modulo_equip, "ModuloEquipamento", "IdeMdl", modulo, "IdeMdl")
modulo_equip = filtrar_fk(
    modulo_equip, "ModuloEquipamento", "IdeSbe", subestacao, "IdeSbe"
)
modulo_manobra = extrair(
    M,
    "ModuloManobra",
    "IdeMdl",
    [
        "IdeSbe",
        "IdeTipAnjManobra",
        "IdcTipMdlMno",
        "IdeTipAnj",
        "SglTipAnj",
        "DscTipAnj",
        "DscMdlMno",
        "QtdDijMdlMno",
        "QtdChvSccMdlMno",
    ],
)
modulo_manobra = filtrar_fk(modulo_manobra, "ModuloManobra", "IdeMdl", modulo, "IdeMdl")
modulo_manobra = filtrar_fk(
    modulo_manobra, "ModuloManobra", "IdeSbe", subestacao, "IdeSbe"
)
# Monta a relação N:M entre linhas e subestações, distinguindo origem e destino.
# A deduplicação é necessária porque uma linha pode aparecer em vários módulos.
ori = L[["IdeLinTms", "IdeSbeOri"]].rename(columns={"IdeSbeOri": "IdeSbe"}).copy()
ori["Papel"] = "Origem"
dst = L[["IdeLinTms", "IdeSbeDsn"]].rename(columns={"IdeSbeDsn": "IdeSbe"}).copy()
dst["Papel"] = "Destino"
linha_sub = pd.concat([ori, dst], ignore_index=True).drop_duplicates()
chave_dupla = linha_sub.groupby(["IdeLinTms", "Papel"], dropna=False)["IdeSbe"].nunique(
    dropna=True
)
problematicas = chave_dupla[chave_dupla > 1].index
if len(problematicas):
    ids = set(x[0] for x in problematicas)
    registrar_rejeicao(
        "LinhaSubestacao",
        "mais de uma subestação para o mesmo papel",
        linha_sub[linha_sub.IdeLinTms.isin(ids)],
    )
    linha_sub = linha_sub[~linha_sub.IdeLinTms.isin(ids)]
linha_sub = filtrar_fk(linha_sub, "LinhaSubestacao", "IdeLinTms", linha, "IdeLinTms")
linha_sub = filtrar_fk(linha_sub, "LinhaSubestacao", "IdeSbe", subestacao, "IdeSbe")
# Mantém somente linhas com dois terminais distintos: uma origem e um destino.
n_term = linha_sub.groupby("IdeLinTms")["IdeSbe"].nunique()
curtos = set(n_term[n_term < 2].index)
if curtos:
    registrar_rejeicao(
        "LinhaSubestacao",
        "terminais não distintos ou incompletos",
        linha_sub[linha_sub.IdeLinTms.isin(curtos)],
    )
    linha_sub = linha_sub[~linha_sub.IdeLinTms.isin(curtos)]
# Reúne as relações prontas para conversão de tipos e inserção no SQLite.
TABELAS = {
    "Agente": agente,
    "Contrato": contrato,
    "Empreendimento": empreendimento,
    "Modulo": modulo,
    "Obra": obra,
    "Subestacao": subestacao,
    "LinhaTransmissao": linha,
    "ModuloLinha": modulo_linha,
    "ModuloEquipamento": modulo_equip,
    "ModuloManobra": modulo_manobra,
    "LinhaSubestacao": linha_sub,
}


# Converte números e datas para os tipos aceitos pelo SQLite.
# Valores inválidos tornam-se NULL e são registrados na auditoria.
def numero(s, inteiro=False):
    src = s.astype("string").str.strip().replace({"": pd.NA})

    # Aceita formatos brasileiros e valores que já estejam no padrão internacional.
    def parse(v):
        if pd.isna(v):
            return None
        v = str(v).strip().replace(" ", "")
        if "," in v:
            v = v.replace(".", "").replace(",", ".")
        try:
            n = float(v)
            if not pd.notna(n):
                return None
            if inteiro and not n.is_integer():
                return None
            return int(n) if inteiro else n
        except:
            return None

    return src.map(parse)


NUMERICOS = {
    "Obra": {"NumVdaUtlMdl": True, "VlrHisRct": False},
    "ModuloLinha": {"NumCcuLinTms": True, "QtdTorLinTms": True, "NumEtnLinTms": False},
    "ModuloEquipamento": {
        "MdaPotAtvMdlEqp": False,
        "MdaTnsScdMdlEqp": False,
        "MdaTnsTiaMdlEqp": False,
        "MdaCpeRaiNegMdlEqp": False,
        "MdaCpeRaivPosMdlEqp": False,
    },
    "ModuloManobra": {"QtdDijMdlMno": True, "QtdChvSccMdlMno": True},
    "LinhaTransmissao": {"NumTensaoBaseLinhaTransm": False},
}
DATAS = {
    "Contrato": ["DatAsnCcd", "DatFimCcd"],
    "Empreendimento": ["DatCaoCgmAtoLgl", "DatOprComEpd", "DatEfeOprComEpd"],
    "Obra": ["DatOprComObr"],
}
# Aplica as conversões configuradas para cada relação e cada coluna.
for nome, df in TABELAS.items():
    for c, inteiro in NUMERICOS.get(nome, {}).items():
        if c not in df:
            continue
        convertido = numero(df[c], inteiro)
        inval = df[c].notna() & convertido.isna()
        registrar_rejeicao(
            nome,
            f"{c}: valor numérico inválido (atributo convertido para NULL)",
            df[inval],
        )
        df[c] = convertido
    for c in DATAS.get(nome, []):
        if c not in df:
            continue
        origem = df[c].copy()
        parsed = pd.to_datetime(origem, dayfirst=True, format="mixed", errors="coerce")
        inval = origem.notna() & parsed.isna()
        registrar_rejeicao(
            nome, f"{c}: data inválida (atributo convertido para NULL)", df[inval]
        )
        df[c] = parsed.dt.strftime("%Y-%m-%d").astype("string")
# Define o esquema relacional, incluindo tipos, chaves e restrições de integridade.
DDL = """
PRAGMA foreign_keys = ON;
CREATE TABLE Agente (IdAgente TEXT PRIMARY KEY,NumCNPJ TEXT,DscRazaoSocial TEXT,NumCEPAgente TEXT,SigUF TEXT,NomMunicipio TEXT,NomBairroAgente TEXT,DscEnderecoAgente TEXT);
CREATE TABLE Contrato (IdeCcd TEXT PRIMARY KEY,IdAgente TEXT,IdcTipoCcd TEXT,NumCnaCcd TEXT,DatAsnCcd TEXT,DatFimCcd TEXT,IdeCcoTarReceita TEXT,FOREIGN KEY(IdAgente) REFERENCES Agente(IdAgente));
CREATE TABLE Empreendimento (IdeEpd TEXT PRIMARY KEY,IdeCcd TEXT NOT NULL,IdeOnsEpd TEXT,NomEpd TEXT,DscEpd TEXT,DscSituacaoEpd TEXT,DatCaoCgmAtoLgl TEXT,DatOprComEpd TEXT,DatEfeOprComEpd TEXT,FOREIGN KEY(IdeCcd) REFERENCES Contrato(IdeCcd));
CREATE TABLE Modulo (IdeMdl TEXT PRIMARY KEY,IdeOnsMdl TEXT,NomMdl TEXT,DscMdl TEXT,IdeTipMdl TEXT,DscSituacaoModulo TEXT,DscTipMdl TEXT,SigTipMdl TEXT,DscClfMdl TEXT,SglClfMdl TEXT,DscItmClfMdl TEXT);
CREATE TABLE Obra (IdeObr TEXT PRIMARY KEY,IdeEpd TEXT NOT NULL,IdeMdl TEXT NOT NULL,DscObr TEXT,DscSitObr TEXT,DscTipObr TEXT,NumVdaUtlMdl INTEGER,VlrHisRct REAL,DatOprComObr TEXT,FOREIGN KEY(IdeEpd) REFERENCES Empreendimento(IdeEpd),FOREIGN KEY(IdeMdl) REFERENCES Modulo(IdeMdl));
CREATE TABLE Subestacao (IdeSbe TEXT PRIMARY KEY,IdeOnsSbe TEXT,NomSubestacao TEXT,NomLongoSubestacao TEXT,SigUFSubestacao TEXT);
CREATE TABLE LinhaTransmissao (IdeLinTms TEXT PRIMARY KEY,NomLinTms TEXT,DscSitLinTms TEXT,NumTensaoBaseLinhaTransm REAL);
CREATE TABLE ModuloLinha (IdeMdl TEXT PRIMARY KEY,IdeLinTms TEXT NOT NULL,NumCcuLinTms INTEGER,QtdTorLinTms INTEGER,NumEtnLinTms REAL,FOREIGN KEY(IdeMdl) REFERENCES Modulo(IdeMdl),FOREIGN KEY(IdeLinTms) REFERENCES LinhaTransmissao(IdeLinTms));
CREATE TABLE ModuloEquipamento (IdeMdl TEXT PRIMARY KEY,IdeSbe TEXT NOT NULL,DscTipEqp TEXT,NumFasMdlEqp TEXT,MdaPotAtvMdlEqp REAL,MdaTnsScdMdlEqp REAL,MdaTnsTiaMdlEqp REAL,DscFinMdlEqp TEXT,MdaCpeRaiNegMdlEqp REAL,MdaCpeRaivPosMdlEqp REAL,DscIdcExsMdlCnr TEXT CHECK (DscIdcExsMdlCnr IN ('S','N') OR DscIdcExsMdlCnr IS NULL),FOREIGN KEY(IdeMdl) REFERENCES Modulo(IdeMdl),FOREIGN KEY(IdeSbe) REFERENCES Subestacao(IdeSbe));
CREATE TABLE ModuloManobra (IdeMdl TEXT PRIMARY KEY,IdeSbe TEXT NOT NULL,IdeTipAnjManobra TEXT,IdcTipMdlMno TEXT,IdeTipAnj TEXT,SglTipAnj TEXT,DscTipAnj TEXT,DscMdlMno TEXT,QtdDijMdlMno INTEGER,QtdChvSccMdlMno INTEGER,FOREIGN KEY(IdeMdl) REFERENCES Modulo(IdeMdl),FOREIGN KEY(IdeSbe) REFERENCES Subestacao(IdeSbe));
CREATE TABLE LinhaSubestacao (IdeLinTms TEXT NOT NULL, IdeSbe TEXT NOT NULL, Papel TEXT NOT NULL CHECK (Papel IN ('Origem','Destino')), PRIMARY KEY(IdeLinTms,Papel), UNIQUE(IdeLinTms,IdeSbe),FOREIGN KEY(IdeLinTms) REFERENCES LinhaTransmissao(IdeLinTms),FOREIGN KEY(IdeSbe) REFERENCES Subestacao(IdeSbe));
"""
# Salva o esquema SQL para documentação e reprodução da estrutura do banco.
(SAIDA / "esquema.sql").write_text(DDL, encoding="utf-8")
# Recria somente o banco de saída; os arquivos CSV de entrada não são modificados.
if DB.exists():
    DB.unlink()
# Abre a conexão e cria as tabelas definidas no DDL.
con = sqlite3.connect(DB)
con.executescript(DDL)
try:
    for nome, df in TABELAS.items():
        # O SQLite recebe None no lugar de NA/NaN, preservando os demais valores.
        valores = [
            tuple(None if pd.isna(x) else x for x in linha)
            for linha in df.itertuples(index=False, name=None)
        ]
        # Monta uma instrução parametrizada com as colunas da relação atual.
        cols = list(df.columns)
        cmd = f'INSERT INTO "{nome}" ({",".join(chr(34) + c + chr(34) for c in cols)}) VALUES ({",".join(["?"] * len(cols))})'
        try:
            con.executemany(cmd, valores)
        except Exception as exc:
            raise RuntimeError(f"Falha ao inserir em {nome}: {exc}") from exc
    # Verifica as chaves estrangeiras antes de confirmar a transação.
    erros = con.execute("PRAGMA foreign_key_check").fetchall()
    assert not erros, f"Falhas de FK: {erros[:10]}"
    con.commit()
finally:
    con.close()
# Exporta os registros rejeitados e o resumo das decisões de limpeza.
# Esses arquivos permitem rastrear por que cada registro foi removido ou convertido.
# Arquivos de auditoria
pd.DataFrame(logs).to_csv(SAIDA / "auditoria.csv", index=False, encoding="utf-8-sig")
for tabela, motivo, df in rejeicoes:
    safe = re.sub(r"[^a-z0-9]+", "_", f"{tabela}_{motivo}".lower()).strip("_")[:95]
    df.to_csv(SAIDA / f"revisar_{safe}.csv", index=False, encoding="utf-8-sig")
print("\nBANCO CRIADO:", DB.resolve())
for nome, df in TABELAS.items():
    print(f"  {nome:22s}: {len(df):>7,} registros")
print("Total de tuplas:", sum(len(t) for t in TABELAS.values()))
print("Integridade referencial: OK (PRAGMA foreign_key_check)")
print("Saídas:", SAIDA.resolve())
```
